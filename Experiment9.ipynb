```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('/content/image.jpg', cv2.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image not found!")
    exit()

def intensity_slicing(img, lower, upper, high_value=255, background=0):
    sliced = np.copy(img)

    mask = (img >= lower) & (img <= upper)
    sliced[mask] = high_value
    sliced[~mask] = background

    return sliced

def intensity_slicing_with_bg(img, lower, upper, high_value=255):
    sliced = np.copy(img)
    mask = (img >= lower) & (img <= upper)
    sliced[mask] = high_value
    return sliced

lower_limit = 100
upper_limit = 180

slice_without_bg = intensity_slicing(image, lower_limit, upper_limit)
slice_with_bg = intensity_slicing_with_bg(image, lower_limit, upper_limit)

plt.figure(figsize=(10,6))

plt.subplot(1,3,1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(slice_without_bg, cmap='gray')
plt.title("Slicing Without Background")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(slice_with_bg, cmap='gray')
plt.title("Slicing With Background")
plt.axis('off')

plt.tight_layout()
plt.show()
```
